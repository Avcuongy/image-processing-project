```python
import os
import kagglehub
import shutil
from sklearn.model_selection import train_test_split
import glob
```

```python
SOURCE_DIR = "../../data/Original Images/Original Images"

BASE_OUTPUT_DIR = "../../dataset_face_recognition"
TRAIN_DIR = os.path.join(BASE_OUTPUT_DIR, "train")
TEST_DIR = os.path.join(BASE_OUTPUT_DIR, "test")

if os.path.exists(BASE_OUTPUT_DIR):
    shutil.rmtree(BASE_OUTPUT_DIR)

os.makedirs(TRAIN_DIR, exist_ok=True)
os.makedirs(TEST_DIR, exist_ok=True)

print(f"Save to: {BASE_OUTPUT_DIR}")
print(f"Source folder: {SOURCE_DIR}")
```

```python
classes = sorted(
    [d for d in os.listdir(SOURCE_DIR) if os.path.isdir(os.path.join(SOURCE_DIR, d))]
)
print(f"Total classes (people) found: {len(classes)}\n")

train_count = 0
test_count = 0

for person_name in classes:
    source_person_dir = os.path.join(SOURCE_DIR, person_name)

    images = []
    for ext in ("*.jpg", "*.jpeg", "*.png", "*.JPG", "*.JPEG", "*.PNG"):
        images.extend(glob.glob(os.path.join(source_person_dir, ext)))

    if len(images) == 0:
        continue

    if len(images) > 1:
        train_imgs, test_imgs = train_test_split(images, test_size=0.2, random_state=42)
    else:
        train_imgs, test_imgs = images, []

    target_train_dir = os.path.join(TRAIN_DIR, person_name)
    target_test_dir = os.path.join(TEST_DIR, person_name)

    os.makedirs(target_train_dir, exist_ok=True)
    os.makedirs(target_test_dir, exist_ok=True)

    for img_path in train_imgs:
        shutil.copy(img_path, target_train_dir)
        train_count += 1

    for img_path in test_imgs:
        shutil.copy(img_path, target_test_dir)
        test_count += 1
```

```python
print(f"Total : {len(classes)}")
total = train_count + test_count
print(f"Total images: {total}")
print(f"Images in Train set: {train_count} images (belonging to {len(classes)} people)")
print(
    f"Images in Test set : {test_count} images (belonging to {len([d for d in os.listdir(TEST_DIR) if os.listdir(os.path.join(TEST_DIR,d))])} people)"
)

if total > 0:
    print(
        f"Actual ratio: {train_count/total*100:.2f}% Train / {test_count/total*100:.2f}% Test"
    )
```

```python
train_people_count = 0
train_count = 0
for person_name in os.listdir(TRAIN_DIR):
    person_path = os.path.join(TRAIN_DIR, person_name)
    if not os.path.isdir(person_path):
        continue
    train_people_count += 1
    train_count += len(os.listdir(person_path))

test_people_count = 0
test_count = 0
for person_name in os.listdir(TEST_DIR):
    person_path = os.path.join(TEST_DIR, person_name)
    if not os.path.isdir(person_path):
        continue
    test_people_count += 1
    test_count += len(os.listdir(person_path))

total = train_count + test_count

print(f"Total images: {total}")
print(f"Number of images in Train set: {train_count} images | {train_people_count} people")
print(f"Number of images in Test set: {test_count} images | {test_people_count} people")

if total > 0:
    print(
        f"Actual ratio: {train_count/total*100:.2f}% Train / {test_count/total*100:.2f}% Test"
    )
else:
    print("Actual ratio: N/A because no images were found in the source directory.")
```
